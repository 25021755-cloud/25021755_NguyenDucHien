```python
#1
def binary_search(arr,left,right,target):
    while left <= right:
        mid = (left + right)//2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
arr = list(map(int,(input().split())))
target = int(input())
left , right = 0 , len(arr) - 1
print(binary_search(arr,left,right,target))
```

```python
#2
def count_occurrences(arr, x):
    n = 0
    for k in arr:
        if k == x:
            n += 1
    return n
arr = list(map(int,(input().split())))
x = int(input())
print(count_occurrences(arr, x))
```

```python
#3
def bubble_sort(arr):
    for i in range (len(arr)):
        swap = False
        for j in range (len(arr)-i-1):
            if arr[j+1] < arr[j]:
                arr[j+1] , arr[j] = arr[j] , arr[j+1]
                swap = True
        if swap == False:
            break
    sn = " ".join(arr)
    return sn
arr = list(map(str,(input().split())))
print(bubble_sort(arr))
def selection_sort(arr2):
    for k in range (len(arr2)):
        min_index = k
        for l in range (k+1 , len(arr2)):
            if arr2[l] < arr2[min_index]:
                min_index = l
        arr2[k] , arr2[min_index] = arr2[min_index] , arr2[k]
    sn2 = " ".join(arr2)
    return sn2
arr2 = list(map(str,(input().split())))
print(selection_sort(arr2))
```

```python
#7
nums = list(map(int,(input().split())))
n = len(nums)
dp = [1] * n
for i in range(n):
    for j in range(i):
        if nums[j] < nums[i]:
            dp[i] = max(dp[i], dp[j] + 1)
print(max(dp))
```

```python
#8
interval = eval(input())
queries = eval(input())
L = []
kq = []
for i in range (len(interval)):
    r = interval[i][1]
    l = interval[i][0]
    d = r - l + 1
    L.append(d)
for j in range (len(queries)):
    tm = []
    for k in range (len(interval)):
        if interval[k][0] <= queries[j] <= interval[k][1]:
            tm.append(L[k])
    if tm == []:
        kq.append(-1)
    else:
        kq.append(min(tm))
print(kq)
```
